```python
def is_even(x):
    if x % 2 == 1:
        print("Odd")
    else:
        print("Even")
is_even(7)
```

```python
def count_vowels(text):
    vowels = ['a', 'e', 'i', 'o', 'u']
    count = 0

    for letter in text.lower():
        if letter in vowels:
            count = count + 1
    return count

count_vowels("Ishmael")
```

```python
def append_item(item, my_list=[]):
    my_list.append(item)
    return my_list

append_item(2, [1,2])
```

```python
words = ["eat", "tea", "tan", "ate", "nat", "bat"]
groups = {}

for word in words:
    key = "".join(sorted(word))
    # print(key)

    if key not in groups:
        groups[key] = []
    # print(groups)

    groups[key].append(word)

result = list(groups.values())
print(result)    
```
