```python
def is_even(x):
    if x % 2 == 1:
        print("Odd")
    else:
        print("Even")
is_even(7)
```

```python
def count_vowels(text):
    vowels = ['a', 'e', 'i', 'o', 'u']
    count = 0

    for letter in text.lower():
        if letter in vowels:
            count = count + 1
    return count

count_vowels("Ishmael")
```
